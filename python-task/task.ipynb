```python
# ① Python 코드에서 `1.01.house.price.csv` 파일을 읽습니다.
# ② 읽은 행을 `df_house` 데이터프레임에 저장합니다.
# ③ 첫 번째 행과 전체 데이터 행 수를 출력합니다.
# ④ 첫 번째 행의 `price` 값이 어떤 자료형인지 출력합니다.
# ⑤ 파일이 존재하지 않으면 `FileNotFoundError`를 처리하여 “파일명과 업로드 여부를 확인하세요.”를 출력합니다.
# ⑥ 파일 읽기에 성공한 경우에만 이후 데이터 확인을 수행하고, 정상 파일명과 존재하지 않는 파일명을 각각 실행하여 결과를 확인합니다. (정상 처리 결과와 오류 처리 결과 제출)

import pandas as pd

def read_file(file):
    try:
        df_house = pd.read_csv(file)
        print(df_house.iloc[0])
        print(df_house.shape[0])
        print(type(df_house["price"].iloc[0]))
    except FileNotFoundError:
        print("파일명과 업로드 여부를 확인하세요.")

read_file("./1.01.house.price.csv")
read_file("./not-exist-file.csv")
```

```python
# ① 빈 리스트 `houses`를 생성합니다.
# ② 반복문을 사용하여 각 행을 순회합니다.
# ③ 각 행의 값을 아래 자료형으로 변환하여 `houses`에 저장합니다.
# ④ 필수 값이 누락되었거나 숫자로 변환할 수 없는 행은 `houses`에 저장하지 않고, 행 위치와 제외 사유를 출력합니다.
# ⑤ 한 행에서 변환 오류가 발생해도 반복문을 중단하지 않고 다음 행을 계속 처리한 뒤, 정상 저장 행 수와 제외 행 수를 출력합니다.

df_house = pd.read_csv("./1.01.house.price.csv")

houses = []
excluded_count = 0

for index, row in df_house.iterrows():
    try:
        if row.isnull().any():
            raise ValueError("필수 값 누락")

        house = {
            "ID": int(row["ID"]),
            "size": int(row["size"]),
            "year": int(row["year"]),
            "price": float(row["price"])
        }

        houses.append(house)

    except Exception as e:
        excluded_count += 1
        print("행 위치:", index)
        print("제외 사유:", e)

print("정상 저장 행 수:", len(houses))
print("제외 행 수:", excluded_count)
```

```python
# ① 원본 `houses`를 유지한 채 `houses_test`를 복사하고, 아래 테스트 데이터를 추가합니다.
# ② `houses_test`를 `for`문으로 반복 처리하고, 조건을 비교하기 전에 `size`·`price` 키와 값을 검사합니다. 키 누락, `None`·결측값, 숫자 변환 불가 값은 각각 제외 사유를 출력하고 건너뜁니다.
# ③ 조건을 비교하기 전에 `size`와 `price`가 존재하고 사용할 수 있는 값인지 검사하세요. 
# ④ `size`가 0 이하이거나 `price`가 음수인 주택은 제외합니다.
# ③ 유효하지 않은 주택을 제외할 때 `or`와 `continue`를 사용합니다.
# ④ 아래 두 조건을 모두 만족하는 주택을 `selected` 리스트에 저장합니다.
# ⑥ 선택한 주택의 ID를 하나씩 출력합니다.
# ⑦ 선택한 주택의 전체 건수를 출력합니다.

houses_test = houses.copy()

houses_test.append({"ID": 101, "year": 5, "price": 3})
houses_test.append({"ID": 102, "size": 35, "year": 5, "price": None})
houses_test.append({"ID": 103, "size": "오류", "year": 5, "price": 3})
houses_test.append({"ID": 104, "size": 35, "year": 5, "price": 3})

selected = []

for house in houses_test:
    if "size" not in house or "price" not in house:
        print(house["ID"], "제외 사유: 필수 키 누락")
        continue

    if pd.isnull(house["size"]) or pd.isnull(house["price"]):
        print(house["ID"], "제외 사유: 필수 값 누락")
        continue

    try:
        size = float(house["size"])
        price = float(house["price"])
    except Exception:
        print(house["ID"], "제외 사유: 숫자 변환 불가")
        continue

    if size <= 0 or price < 0:
        print(house["ID"], "제외 사유: 유효하지 않은 값")
        continue

    if size >= 30 and price <= 4:
        selected.append(house)

for house in selected:
    print("선택한 주택 ID:", house["ID"])

print("선택한 주택 수:", len(selected))
```

```python
# ① 가격을 크기로 나눈 값을 계산하는 unit_price 함수를 작성합니다.
# ② 숫자 또는 숫자 형태의 문자열을 입력받도록 작성합니다.
# ③ 정상 입력이면 가격을 크기로 나눈 값을 반환합니다.
# ④ 크기가 0 이하이거나 가격이 음수이면 안내 문구를 출력하고 None을 반환합니다.
# ⑤ ValueError가 발생하면 값 오류 안내 문구를 출력하고 None을 반환합니다.
# ⑥ TypeError가 발생하면 자료형 오류 안내 문구를 출력하고 None을 반환합니다.
# ⑦ 잘못된 입력이 있어도 프로그램이 중단되지 않도록 처리합니다.
# ⑧ 전체 houses에 함수를 적용하여 주택 ID와 크기당 가격을 출력합니다.
# ⑨ 테스트 데이터를 순서대로 실행하여 반환값과 안내 문구를 확인합니다.

def unit_price(price, size):
    try:
        price = float(price)
        size = float(size)

        if size <= 0:
            print("크기 오류: 크기는 0보다 커야 합니다.")
            return None

        if price < 0:
            print("가격 오류: 가격은 음수일 수 없습니다.")
            return None

        return price / size

    except ValueError:
        print("값 오류: 숫자로 변환할 수 없는 문자열입니다.")
        return None

    except TypeError:
        print("자료형 오류: 숫자 또는 숫자 형태의 문자열을 입력하세요.")
        return None


for house in houses:
    result = unit_price(house["price"], house["size"])
    print("주택 ID:", house["ID"], "크기당 가격:", result)


print(unit_price(4, 40))
print(unit_price("4", "40"))
print(unit_price(4, 0))
print(unit_price(-4, 40))
print(unit_price("오류", 40))
print(unit_price(None, 40))
print(unit_price(6, 30))
```

```python
# ① `houses`를 활용하여 전체 주택 ID를 원래 순서대로 저장하고 출력합니다.
# ② 주택의 `year` 값을 중복 없이 저장하고 출력합니다.
# ③ 주택 ID로 가격을 조회할 수 있도록 저장하고, ID 3의 가격을 출력합니다.
# ④ 최소 크기 `30`과 최대 가격 `4`를 **각각 하나의 원소로 갖는 묶음**으로 저장하세요. 생성한 뒤 원소를 추가/삭제하거나 다른 값으로 교체할 수 없는 자료구조를 선택하고, 선택 이유를 설명하세요.
# ⑤ 각 작업에 리스트·튜플·딕셔너리·집합 중 적합한 자료구조를 선택하고, 선택 이유 및 선택하지 않은 자료구조 1개와의 차이를 코드 주석으로 작성합니다. 순서 유지, 중복 허용 여부, 키를 통한 조회, 변경 가능 여부 중 해당 작업의 판단 근거를 포함합니다.

# list는 데이터의 순서를 유지하고 중복을 허용하기 때문에 선택
# set은 중복을 허용하지 않고 원래 순서를 보장하지 않는다.
house_ids = []

for house in houses:
    house_ids.append(house["ID"])

print("전체 주택 ID:", house_ids)

# set은 중복된 year 값을 제거할 수 있기 때문에 선택
# list는 중복된 값을 허용
years = set()

for house in houses:
    years.add(house["year"])

print("중복 없는 year:", years)

# dict는 주택 ID를 키로 사용하여 가격을 조회할 수 있기 때문에 선택
# list는 인덱스를 통해 값을 조회하므로 ID를 키로 직접 조회하는 데 적합하지 않다.
house_prices = {}

for house in houses:
    house_prices[house["ID"]] = house["price"]

print("ID 3의 가격:", house_prices.get(3))

# 튜플은 생성한 이후 원소를 추가, 삭제하거나 변경할 수 없기 때문에 선택
# list는 생성한 이후에도 원소를 추가, 삭제하거나 변경할 수 있다.
conditions = (30, 4)

print("최소 크기:", conditions[0])
print("최대 가격:", conditions[1])
```
